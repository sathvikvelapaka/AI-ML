When working with arrays, you often need to know their shape, size, and data type.

**Beginner Explanation:**

We're importing the NumPy library, which gives us powerful array structures and fast mathematical operations.

```python
import numpy as np

# Let's create a 2D array (2 rows, 3 columns)
matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
print("Our Matrix:\n", matrix)
```

**Advanced Explanation:**

NumPy is implemented in C and provides a Python‑level API for high‑performance vectorized operations.

### Inspecting the Array

**Beginner Explanation:**

This code performs a NumPy operation.

```python
print("Dimensions (ndim):", matrix.ndim)  # Should be 2
print("Shape (shape):", matrix.shape)     # Should be (2, 3) meaning 2 rows, 3 cols
print("Total Elements (size):", matrix.size) # Should be 6
print("Data Type (dtype):", matrix.dtype) # Default is int64 (or int32 depending on OS)
```

## 🔹 Extra Array Properties (Beyond the Basics)

```python
import numpy as np
# Example array used for all snippets
arr = np.arange(12).reshape(3, 4)
print('Example array:
', arr)
```

**1️⃣ Shape, ndim, size, dtype**

```python
print('Shape :', arr.shape)
print('ndim  :', arr.ndim)
print('Size  :', arr.size)
print('dtype :', arr.dtype)
```

**2️⃣ itemsize & nbytes**

```python
print('Bytes per element (itemsize):', arr.itemsize)
print('Total bytes used (nbytes)   :', arr.nbytes)
```

**3️⃣ strides**

```python
print('Strides (bytes per step):', arr.strides)
```

**4️⃣ transpose (arr.T)**

```python
print('Original:
', arr)
print('Transposed:
', arr.T)
```

**5️⃣ flat iterator**

```python
print('First 5 flat elements:', list(arr.flat[:5]))
arr.flat[0] = -99
print('After modifying via .flat: first element =', arr.flat[0])
```

**6️⃣ reshape vs resize**

```python
reshaped = arr.reshape((2, 6))
print('Reshaped (view) shape:', reshaped.shape)
arr.resize((2, 6))
print('Resized in‑place shape:', arr.shape)
```

**7️⃣ squeeze**

```python
b = np.arange(3).reshape((1, 3, 1))
print('Before squeeze:', b.shape)
print('After squeeze :', b.squeeze().shape)
```

**8️⃣ copy**

```python
c = arr.copy()
c[0,0] = 999
print('Original unchanged? :', arr[0,0])
print('Copy modified        :', c[0,0])
```

**9️⃣ tolist**

```python
py_list = arr.tolist()
print('Python list version:', py_list)
```

**🔟 flags**

```python
print('Memory flags:', arr.flags)
```

**1️⃣1️⃣ base**

```python
view = arr[::2]
print('view.base is arr? :', view.base is arr)
```

**1️⃣2️⃣ dtype.name**

```python
print('Data type name:', arr.dtype.name)
```
