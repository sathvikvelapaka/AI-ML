# Lesson 2: Creating NumPy Arrays & Visualization

We can create arrays from standard Python lists, or use built-in functions to generate arrays automatically.

**Beginner Explanation:**

We're importing the NumPy library, which gives us powerful array structures and fast mathematical operations.

```python
import numpy as np
import matplotlib.pyplot as plt

def visualize_array(arr, title):
    plt.figure(figsize=(6,2))
    plt.imshow([arr], cmap='Blues', aspect='auto')
    for i in range(len(arr)):
        plt.text(i, 0, str(arr[i]), ha='center', va='center', color='black' if arr[i] < max(arr)/2 else 'white', fontweight='bold')
    plt.title(title)
    plt.yticks([])
    plt.show()
```

**Advanced Explanation:**

NumPy is implemented in C and provides a Python‑level API for high‑performance vectorized operations.

### 1. Simple Array from List

**Beginner Explanation:**

Here we create a NumPy array from a Python list. Arrays store data in a contiguous block of memory, which makes them fast.

```python
arr_simple = np.array([10, 20, 30, 40, 50])
print("Simple Array:", arr_simple)
visualize_array(arr_simple, 'np.array([10, 20, 30, 40, 50])')
```

**Advanced Explanation:**

`np.array` copies the data into a contiguous block of memory (C‑order by default). This enables SIMD optimizations.

### 2. Generating Sequences (`arange` and `linspace`)
- `arange(start, stop, step)`: Works like Python's range.
- `linspace(start, stop, num)`: Gives you exactly `num` evenly spaced points.

**Beginner Explanation:**

`np.arange` generates a sequence of numbers with a given step. It works like Python's `range` but returns an array.

```python
arr_arange = np.arange(0, 10, 2)
print("Arange (0 to 10, step 2):", arr_arange)
visualize_array(arr_arange, 'np.arange(0, 10, 2)')

arr_linspace = np.linspace(0, 1, 5)
print("Linspace (5 points between 0 and 1):", arr_linspace)
visualize_array(arr_linspace, 'np.linspace(0, 1, 5)')
```

**Advanced Explanation:**

Internally it computes `start + step * i` for each index `i`, returning a new view when possible.

### 3. Matrices (2D Arrays)
Let's create matrices full of zeros, full of a specific number, or an Identity matrix.

**Beginner Explanation:**

We create an array filled with zeros. This is useful for initializing a matrix before filling it.

```python
zeros_matrix = np.zeros((3, 3)) # 3 rows, 3 columns of zeros
print("Zeros Matrix:\n", zeros_matrix)

full_matrix = np.full((2, 4), 99) # 2 rows, 4 columns filled with 99
print("\nFull Matrix:\n", full_matrix)

identity_matrix = np.eye(4) # 4x4 matrix with 1s on the diagonal
print("\nIdentity Matrix:\n", identity_matrix)
```

**Advanced Explanation:**

The array is allocated and zero‑filled using `memset`, which is highly efficient.
