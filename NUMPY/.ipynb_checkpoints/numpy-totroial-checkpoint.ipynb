```python
import time
import numpy as np

# 1. Set the data size (1 million items)
size = 100_000_000

# -------------------------------
# Python List (Standard Loop)
# -------------------------------
py_list = list(range(size))

start_time = time.time()
sq_list = [x**2 for x in py_list]
end_time = time.time()

python_duration = end_time - start_time
print(f"Python list time : {python_duration:.6f} seconds")

# -------------------------------
# NumPy Array (Vectorization)
# -------------------------------
np_arr = np.arange(size)

start_time = time.time()
sq_arr = np_arr**2  # vectorized operation
end_time = time.time()

numpy_duration = end_time - start_time
print(f"NumPy array time : {numpy_duration:.6f} seconds")

# -------------------------------
# Performance Difference
# -------------------------------
speedup = python_duration / numpy_duration
print(f"NumPy is {speedup:.1f}x faster")
```

```python


#need heading 


import sys
import numpy as np

# 1. Create a sample size
size = 1000

# Python list memory: size of one element * number of elements + list overhead
py_list = list(range(size))
py_list_size = sys.getsizeof(py_list[0]) * len(py_list) + sys.getsizeof(py_list)

# NumPy array memory: item size * number of elements (or np_arr.nbytes)
np_arr = np.arange(size)
np_arr_size = np_arr.nbytes

print(f"Python list size: {py_list_size} bytes")
print(f"NumPy array size: {np_arr_size} bytes")
```

```python
import numpy as np

# 1. Creating a standard integer NumPy array
# All elements are numbers, so dtype defaults to int (e.g., int64)
arr = np.array([1, 2, 3, 4, 5])

# Access dtype using dot notation: arr.dtype (not as a separate variable)
print(arr, type(arr), arr.dtype)

# 2. Creating a NumPy array with mixed types (integers + string)
# NumPy arrays are homogeneous (all elements must have the same type).
# Adding "prime" forces NumPy to upcast all numbers into Unicode strings (<U21).
arr2 = np.array([1, 2, 3, 4, 5, "prime"])

# Fixed: Changed `arr2, dtype` to `arr2.dtype`
print(arr2, type(arr2), arr2.dtype)
```

```python
import numpy as np

# 1D Array (Vector)
# .ndim returns the number of dimensions (axes) -> 1
# .shape returns a tuple representing elements along each axis -> (5,)
arr_1d = np.array([1, 2, 3, 4, 5])
print("1D Array:")
print(arr_1d)
print("Dimensions (ndim) :", arr_1d.ndim)
print("Shape (shape)     :", arr_1d.shape)

print("-" * 30)

# 2D Array (Matrix - rows, columns)
# .ndim -> 2
# .shape -> (2, 3) meaning 2 rows and 3 columns
arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
print("2D Array:")
print(arr_2d)
print("Dimensions (ndim) :", arr_2d.ndim)
print("Shape (shape)     :", arr_2d.shape)
```

```python
import numpy as np

# ----------------------------------------------------
# 1. np.full()
# ----------------------------------------------------
# np.full(shape, fill_value) creates an array of given shape filled with a constant value.
# Error in your cell: used a dot `.` instead of a comma `,` between shape and value ((2, 3).100 -> (2, 3), 100).
arr_full = np.full((2, 3), 100)
print("np.full result:")
print(arr_full)
print("Shape:", arr_full.shape)

print("-" * 40)

# ----------------------------------------------------
# 2. np.eye()
# ----------------------------------------------------
# np.eye(N) is a function, so it must be called with parentheses ().
# It creates a 2D Identity Matrix of size N x N (1s on the main diagonal, 0s elsewhere).
arr_eye = np.eye(3)  # creates a 3x3 identity matrix
print("np.eye result:")
print(arr_eye)
print("Shape:", arr_eye.shape)
```

```python
import numpy as np

# ----------------------------------------------------
# 1. np.arange(start, stop, step, dtype)
# ----------------------------------------------------
# Returns evenly spaced values within a given interval.
# Parameters:
# - start : Starting value of the sequence (inclusive, default=0)
# - stop  : End value of the sequence (exclusive)
# - step  : Spacing between values (default=1)
# - dtype : Desired output data type (optional)

# Example: Generate numbers from 1 to 10 (stop is 11 so 10 is included)
arr_1_to_10 = np.arange(1, 11)
print("np.arange(1, 11):")
print(arr_1_to_10)

print("-" * 40)

# Example with step size (odd numbers from 1 to 10)
arr_step = np.arange(1, 11, 2)
print("np.arange(1, 11, 2):")
print(arr_step)

print("-" * 40)

# ----------------------------------------------------
# 2. Diagonal Matrix prefilled with values
# ----------------------------------------------------
# np.eye() creates an identity matrix (1s on diagonal).
# Multiply np.eye() by a scalar to prefill diagonal with any custom value.
custom_diag = np.eye(3) * 5
print("3x3 Matrix with 5 on diagonal:")
print(custom_diag)

# Alternatively, np.diag() places specific elements along diagonal:
diag_custom_vals = np.diag([10, 20, 30])
print("\nnp.diag([10, 20, 30]):")
print(diag_custom_vals)
```

```python
import numpy as np

# ----------------------------------------------------
# np.linspace(start, stop, num=50, endpoint=True, retstep=False)
# ----------------------------------------------------
# Generates evenly spaced numbers over a specified interval.
# Key difference from arange: you specify the total count of numbers (num), not the step size.

# Parameters:
# - start    : Starting value of the sequence
# - stop     : End value of the sequence (included by default)
# - num      : Number of evenly spaced samples to generate (default is 50)
# - endpoint : If True (default), stop is the last sample. If False, stop is excluded.
# - retstep  : If True, returns (array, step_size) tuple.

# 1. Basic example: 5 values between 1 and 10
arr_lin = np.linspace(1, 10, 5)
print("np.linspace(1, 10, 5):")
print(arr_lin)

print("-" * 40)

# 2. Excluding the stop value (endpoint=False)
arr_no_endpoint = np.linspace(1, 10, 5, endpoint=False)
print("np.linspace(1, 10, 5, endpoint=False):")
print(arr_no_endpoint)

print("-" * 40)

# 3. Returning the step size (retstep=True)
arr_step, step_val = np.linspace(0, 1, 5, retstep=True)
print("Array with step value:")
print("Values   :", arr_step)
print("Step size:", step_val)
```

```python
import numpy as np

# ----------------------------------------------------
# Essential NumPy Array Properties (Attributes)
# ----------------------------------------------------

# Create a sample 2D array (2 rows, 3 columns)
arr = np.array([[10, 20, 30], [40, 50, 60]], dtype="int32")

print("Array:\n", arr)
print("-" * 40)

# 1. .ndim -> Number of dimensions / axes (1 for 1D, 2 for 2D, etc.)
print("1. ndim (dimensions)      :", arr.ndim)

# 2. .shape -> Tuple of array dimensions (rows, columns)
print("2. shape (rows, cols)     :", arr.shape)

# 3. .size -> Total number of elements in the array
print("3. size (total elements)  :", arr.size)

# 4. .dtype -> Data type of the elements (e.g., int32, int64, float64)
print("4. dtype (data type)      :", arr.dtype)

# 5. .itemsize -> Size of each individual element in bytes
print("5. itemsize (bytes/item)  :", arr.itemsize)

# 6. .nbytes -> Total memory consumed by array elements (size * itemsize)
print("6. nbytes (total bytes)   :", arr.nbytes)
```
