# Lesson 4: Operations, Indexing & Slicing

We can easily grab specific elements from an array, or do math on the entire array at once.

### 1. Basic Math (Vectorization)
No `for` loops needed! Math operations apply to every element automatically.

```python
import numpy as np

arr = np.array([1, 2, 3, 4])
print("Original:", arr)
print("Add 10:  ", arr + 10)
print("Multiply:", arr * 2)
```

### 2. Indexing and Slicing (1D)
Works just like Python lists.

```python
arr = np.arange(10, 100, 10)
print("Array:", arr)  # [10 20 30 40 50 60 70 80 90]
print("First element:", arr[0])
print("Last element:", arr[-1])
print("First 3 elements:", arr[:3])
```

### 3. Reshaping and Slicing Matrices (2D)

```python
# Create a 1D array of 9 elements, then reshape it into a 3x3 matrix
matrix = np.arange(1, 10).reshape((3, 3))
print("3x3 Matrix:\n", matrix)

print("\nElement at row 0, col 2:", matrix[0, 2])
print("Entire first row:", matrix[0, :])
print("Entire first column:", matrix[:, 0])
```
