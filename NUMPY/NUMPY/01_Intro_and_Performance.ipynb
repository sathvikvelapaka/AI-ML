# Lesson 1: Introduction to NumPy

NumPy (Numerical Python) is the core library for scientific computing in Python. It provides a high-performance multidimensional array object, and tools for working with these arrays.

## Why use NumPy?
- **Fast:** Written in C, it stores data in continuous memory blocks.
- **Efficient:** Uses less memory than standard Python lists.
- **Convenient:** Allows mathematical operations directly on whole arrays (Vectorization).

### Speed Comparison: Python List vs NumPy Array
Let's see how much faster NumPy is when we try to square 1 million numbers!

```python
import time
import numpy as np

size = 1_000_000

# 1. Python List
py_list = list(range(size))
start = time.time()
sq_list = [x**2 for x in py_list]
py_time = time.time() - start

# 2. NumPy Array
np_arr = np.arange(size)
start = time.time()
sq_arr = np_arr**2
np_time = time.time() - start

print(f"Python List took: {py_time:.4f} seconds")
print(f"NumPy Array took: {np_time:.4f} seconds")
print(f"NumPy is {py_time / np_time:.1f} times faster!")
```
